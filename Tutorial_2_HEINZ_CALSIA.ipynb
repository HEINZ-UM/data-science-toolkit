5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n = 2
p = 3
m = 4

array1 = np.random.randint(0, 10, size=(n,p))
array2 = np.random.randint(0,10, size=(p,m))

result = np.dot(array1, array2)

print(result)
```
